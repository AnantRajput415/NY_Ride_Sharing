**Observe Combined data**

```python
output_file = "../combinedrids/combined_unique_rid.parquet" # change accordingly
combined_df = pd.read_parquet(output_file, engine="pyarrow")
combined_df
```

```python
combined_df.nunique(dropna=True)
```

```python
print(combined_df['mode'].value_counts())
print(combined_df['rideStatus'].value_counts())
print(combined_df['status'].value_counts())
```

```python
combined_df = combined_df[combined_df['status'] == "COMPLETED"]
combined_df = combined_df[combined_df['mode'] == "ONLINE"]
```

```python
combined_df.nunique(dropna=True)
```

**Preprocess Combined data**

```python
cols_to_keep = ['hashed_rid', 'hashed_driver_id', 'trip_start_lat', 'trip_start_lon', 'trip_end_lat', 'trip_end_lon', 'trip_start_time', 'trip_end_time', 'chargeable_distance', 'fare', 'traveled_distance','rider_id']
combined_df = combined_df[cols_to_keep]
```

```python
combined_df = combined_df.dropna()
```

```python
combined_df["chargeable_distance"]
```

```python
combined_df.nunique(dropna=True)
```

Save new combine data with required columns

```python
OUTPUT_FILE = Path("./combinedrids/combined_unique_hashed_rid.parquet")
combined_df.to_parquet(OUTPUT_FILE, index=False)

print(f"Saved {len(combined_df)} rows to {OUTPUT_FILE}")
```
